```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.datasets import fashion_mnist
```

# Load the dataset

[View Dataset](https://keras.io/api/datasets/fashion_mnist/)

```python
(xtrain, ytrain), (xtest, ytest) = fashion_mnist.load_data()
```

```python
xtrain.shape
```

```python
xtest.shape
```

```python
plt.imshow(xtrain[100])
```

# Normalize the pixel values

```python
xtrain = xtrain / 255.0
xtest = xtest / 255.0
```

```python
xtest[20]
```

# One-hot encode the labels

```python
y_train = tf.keras.utils.to_categorical(ytrain, 10)
y_test = tf.keras.utils.to_categorical(ytest, 10)
```

```python
y_train
```

```python
y_test
```

```python
y_test.shape
```

# Artificial Neural Network (ANN)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

# Reshape the data to be flat (ANN works with 1D data)

```python
x_train_ann = xtrain.reshape(-1, 28*28)
x_test_ann = xtest.reshape(-1, 28*28)
```

# Model

```python
ann_model = Sequential([
    Dense(128, activation='relu', input_shape=(28*28,)),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(10, activation='softmax') # FOR Binary -> 'sigmoid'
])
```

# Compile the model

[List of Optimizers](https://keras.io/api/optimizers/) <br>
[Metrics](https://keras.io/api/metrics/accuracy_metrics/) <br>
[Loss](https://keras.io/api/losses/)

```python
ann_model.compile( optimizer='adam', 
                   loss='categorical_crossentropy', 
                   metrics=['accuracy']
                 )
```

# Train the model

```python
ann_history = ann_model.fit(x_train_ann, 
                            y_train, 
                            batch_size=50,
                            epochs=10, 
                            validation_data=(x_test_ann, y_test))

ann_history
```

# Evaluate the model

```python
ann_test_loss, ann_test_acc = ann_model.evaluate(x_test_ann, y_test)
print(f'ANN Test Accuracy: {ann_test_acc}')
```

# Convolutional Neural Network (CNN)

```python
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten
```

# Reshape the data to be 3D for CNN (28x28x1)

Reshape the input data to (28, 28, 1) since CNNs require 3D input. `(28, 28, 1)` represent the image height, width, and the number of color channels (1 for grayscale). The `-1` eliminates the need to manually calculate the batch size, which makes the code more flexible. If the number of samples changes, the -1 ensures the reshape function adapts accordingly.

```python
x_train_cnn = xtrain.reshape(-1, 28, 28, 1)
x_test_cnn = xtest.reshape(-1, 28, 28, 1)
```

# CNN model

- **Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1))**: First convolutional layer with 32 filters of size 3x3, applied to 28x28 grayscale images. ReLU activation introduces non-linearity to help model learn complex patterns.

- **MaxPooling2D((2,2))**: Downsamples the feature maps by taking the maximum value in each 2x2 window, reducing the spatial dimensions while retaining important features.

- **Conv2D(64, (3,3), activation='relu')**: Second convolutional layer with 64 filters of size 3x3 to extract more complex features from the data.

- **MaxPooling2D((2,2))**: Further downsampling using a 2x2 window to reduce the spatial dimensions after the second convolutional layer.

- **Flatten()**: Flattens the 2D feature maps from the convolutional layers into a 1D vector, preparing it for the fully connected layers.

- **Dense(128, activation='relu')**: Fully connected (dense) layer with 128 neurons, which learns to combine features extracted by previous layers. ReLU activation helps in learning complex representations.

- **Dense(10, activation='softmax')**: Output layer with 10 neurons (one for each class), using softmax activation to output class probabilities for classification.


```python
cnn_model = Sequential([
    
    Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    MaxPooling2D((2,2)),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D((2,2)), #cnn er feature extraction
    
    # Now ANN
    Flatten(), #1D
    Dense(128, activation='relu'), 
    Dense(10, activation='softmax')
    
])
```

```python
# Compile the model
cnn_model.compile(optimizer='adam', 
                  loss='categorical_crossentropy', 
                  metrics=['accuracy'])
```

```python
# Train the model
cnn_history = cnn_model.fit(x_train_cnn, 
                            y_train, epochs=5, 
                            validation_data=(x_test_cnn, y_test))
```

# Evaluate the model

```python
cnn_test_loss, cnn_test_acc = cnn_model.evaluate(x_test_cnn, y_test)
print(f'CNN Test Accuracy: {cnn_test_acc}')
```

# Transfer Learning using VGG19 for the Fashion MNIST dataset.

[Transfer Learning](https://keras.io/api/applications/)

Since `VGG19` was trained on the `ImageNet dataset`, it expects RGB images (3 channels). The Fashion MNIST dataset consists of grayscale images (1 channel), so we'll need to convert the single-channel `grayscale images` to three-channel `RGB images` before feeding them to the VGG19 model.

```python
from tensorflow.keras.applications import VGG19
from tensorflow.keras.layers import GlobalAveragePooling2D, Dropout, Dense
from tensorflow.keras.models import Sequential
```

# Resize the images to 32x32

VGG19 was trained on the ImageNet dataset, expects a minimum input size of 32x32 pixels, but the Fashion MNIST images are 28x28 pixels. We can resize the input images from 28x28 to at least 32x32 before passing them to VGG19.

```python
x_train_resized = tf.image.resize(x_train_cnn, [32, 32])
x_test_resized = tf.image.resize(x_test_cnn, [32, 32])
```

```python
x_train_resized.shape
```

# Convert grayscale images to RGB

```python
x_train_rgb = tf.image.grayscale_to_rgb(x_train_resized)
x_test_rgb = tf.image.grayscale_to_rgb(x_test_resized)
```

```python
plt.imshow(x_train_rgb[100])
```

# Load the VGG19 model without the top layers

```python
base_model = VGG19(input_shape=(32, 32, 3), include_top=False, weights='imagenet')
```

# Freeze the layers in the VGG19 model

Freezing the layers allows you to leverage the power of transfer learning by keeping the pre-trained features and only training the additional task-specific layers, which is especially beneficial for smaller datasets or `limited computational resources`.

```python
base_model.trainable = False
```

# Build the transfer learning model

```python
transfer_model_vgg19 = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dropout(0.2),
    Dense(10, activation='softmax')  # Output layer for 10 classes
])
```

# Compile the model

```python
transfer_model_vgg19.compile(optimizer='adam', 
                             loss='categorical_crossentropy', 
                             metrics=['accuracy'])
```

# Train the model

```python
transfer_history_vgg19 = transfer_model_vgg19.fit(x_train_rgb, 
                                                  y_train, epochs=3, # 100 
                                                  validation_data=(x_test_rgb, y_test))
```

# Evaluate the model on the test set

```python
vgg19_test_loss, vgg19_test_acc = transfer_model_vgg19.evaluate(x_test_rgb, y_test)
print(f'VGG19 Transfer Learning Test Accuracy: {vgg19_test_acc}')
```

# Test accuracies

```python
test_accuracies = [ann_test_acc, cnn_test_acc, vgg19_test_acc]
test_losses = [ann_test_loss, cnn_test_loss, vgg19_test_loss]
models = ['ANN', 'CNN', 'VGG19']

# Bar plot for accuracy comparison
plt.figure(figsize=(10, 5))

# Accuracy plot
plt.subplot(1, 2, 1)
plt.bar(models, test_accuracies, color=['blue', 'green', 'red'])
plt.title('Test Accuracy Comparison')
plt.xlabel('Models')
plt.ylabel('Accuracy')

# Loss plot
plt.subplot(1, 2, 2)
plt.bar(models, test_losses, color=['blue', 'green', 'red'])
plt.title('Test Loss Comparison')
plt.xlabel('Models')
plt.ylabel('Loss')

plt.tight_layout()
plt.show()
```
