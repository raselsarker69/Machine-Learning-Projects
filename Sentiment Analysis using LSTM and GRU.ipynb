```python
import tensorflow as tf
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb
```

# Load the dataset

The IMDb movie review sentiment classification task is a NLP project where the goal is to classify a given movie review as `positive` or `negative` based on its content. The Long Short-Term Memory (LSTM) and Gated Recurrent Unit (GRU) networks are well-suited for sequence data like text.

```python
vocab_size = 10000  # Keep only the top 10k words
maxlen = 250  # Maximum length of the reviews

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# Pad sequences to ensure all reviews are the same length
X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)
```

```python
print(f'Training data shape: {X_train.shape}')
print(f'Testing data shape: {X_test.shape}')
```

# Build LSTM Model

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
```

```python
lstm_model = Sequential()
lstm_model.add(Embedding(vocab_size, 128, input_length=maxlen))
lstm_model.add(LSTM(128, dropout=0.2, recurrent_dropout=0.2))
lstm_model.add(Dense(1, activation='sigmoid')) #output layer

# Compile the model
lstm_model.compile(loss='binary_crossentropy',
                   optimizer='adam', metrics=['accuracy'])
```

# Display the model summary

```python
lstm_model.summary()
```

# Train LSTM model

```python
history_lstm = lstm_model.fit(X_train, y_train, 
                              epochs=3, 
                              batch_size=64, 
                              validation_data=(X_test, y_test))
```

```python
# Evaluate LSTM
lstm_loss, lstm_acc = lstm_model.evaluate(X_test, y_test, verbose=2) 
print(f'LSTM Test Accuracy: {lstm_acc:.4f}')
```

# Build GRU model

```python
from tensorflow.keras.layers import GRU
```

```python
gru_model = Sequential()
gru_model.add(Embedding(vocab_size, 128, input_length=maxlen))
gru_model.add(GRU(128, dropout=0.2, recurrent_dropout=0.2))
gru_model.add(Dense(1, activation='sigmoid'))

# Compile the model
gru_model.compile(loss='binary_crossentropy',
                  optimizer='adam', metrics=['accuracy'])

# Display the model summary
gru_model.summary()
```

# Train GRU model

```python
history_gru = gru_model.fit(X_train, y_train, 
                            epochs=2, 
                            batch_size=64, 
                            validation_data=(X_test, y_test))
```

```python
gru_loss, gru_acc = gru_model.evaluate(X_test, y_test, verbose=2)
print(f'GRU Test Accuracy: {gru_acc:.4f}')
```

# Function to plot accuracy and loss

```python
import matplotlib.pyplot as plt
```

```python
def plot_history(history, title):
    plt.plot(history.history['accuracy'])
    plt.plot(history.history['val_accuracy'])
    plt.title(f'{title} Accuracy')
    plt.ylabel('Accuracy')
    plt.xlabel('Epoch')
    plt.legend(['train', 'validation'], loc='upper left')
    plt.show()
```

```python
# Plot for LSTM
plot_history(history_lstm, "LSTM")
```

```python
# Plot for GRU
plot_history(history_gru, "GRU")
```
