  <h2 style="text-align: center; color: #333366;">Removing outliers using the IQR Method</h2>

```python
# Removing outliers using the IQR method
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)  # First quartile (25th percentile)
    Q3 = df[column].quantile(0.75)  # Third quartile (75th percentile)
    IQR = Q3 - Q1  # Interquartile range
    lower_bound = Q1 - 1.5 * IQR  # Lower threshold
    upper_bound = Q3 + 1.5 * IQR  # Upper threshold
    
    # Filter out outliers
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
```

```python
# Apply outlier removal to each numerical column
for column in numerical_columns:
    df = remove_outliers(df, column)
```

```python
from sklearn.model_selection import train_test_split

# Assuming 'target' is your target column (update as per your dataset)
target_column = 'CO2_emissions'  # For regression
classification_target = 'fuel_type'  # For classification (example)

# Regression
X_reg = df.drop(columns=[target_column])
y_reg = df[target_column]

# Classification
X_clf = df.drop(columns=[classification_target])
y_clf = df[classification_target]

# Split the data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_clf, y_clf, test_size=0.2, random_state=42)
```

  <h2 style="text-align: center; color: #333366;">Regression </h2>

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor

# Initialize models
regressors = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(),
    "Random Forest": RandomForestRegressor(),
    "Gradient Boosting": GradientBoostingRegressor(),
    "XGBoost": XGBRegressor()
}

# Train models
for name, model in regressors.items():
    model.fit(X_train_reg, y_train_reg)
    print(f"{name} model trained.")
```

  <h2 style="text-align: center; color: #333366;">Classification </h2>

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier

# Initialize models
classifiers = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "XGBoost": XGBClassifier()
}

# Train models
for name, model in classifiers.items():
    model.fit(X_train_clf, y_train_clf)
    print(f"{name} model trained.")
```

<div style="background-color: #f9f9fc; color: #333366; border-radius: 12px; margin: 20px auto; padding: 20px; border: 2px solid #4c4cff; max-width: 1000px; font-family: Arial, sans-serif; line-height: 1.6;">
  <h2 style="text-align: center; color: #333366;">Regression metrics</h2>

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

for name, model in regressors.items():
    y_pred = model.predict(X_test_reg)
    print(f"{name} Evaluation:")
    print(f"MAE: {mean_absolute_error(y_test_reg, y_pred)}")
    print(f"MSE: {mean_squared_error(y_test_reg, y_pred)}")
    print(f"R2 Score: {r2_score(y_test_reg, y_pred)}\n")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

for name, model in classifiers.items():
    y_pred = model.predict(X_test_clf)
    print(f"{name} Evaluation:")
    print(f"Accuracy: {accuracy_score(y_test_clf, y_pred)}")
    print(f"Precision: {precision_score(y_test_clf, y_pred)}")
    print(f"Recall: {recall_score(y_test_clf, y_pred)}")
    print(f"F1 Score: {f1_score(y_test_clf, y_pred)}")
    print(f"Confusion Matrix:\n{confusion_matrix(y_test_clf, y_pred)}\n")
```
